# 01. Data Overview & Analytical Framing
## データ概要と分析の枠組み

## 1. Analytical Objective / 分析目的

The objective of this analysis is to understand how Japanese pear production
is structurally supported across different prefectures.
Rather than focusing on short-term fluctuations,
the analysis examines whether production is primarily sustained by
cultivated area, yield efficiency, or a combination of both.

This perspective reflects practical questions faced by production regions,
such as how current production levels are maintained
and how structural changes may affect future sustainability.

本分析の目的は、日本梨の生産が都道府県ごとに
どのような構造によって支えられているのかを明らかにすることです。

年ごとの短期的な変動ではなく、
生産量が「結果樹面積」に依存しているのか、
それとも「単位収量（効率）」によって支えられているのか、
あるいはその両方なのかという
生産構造の違いに着目しています。

これは、産地が直面する
「現在の生産水準は何によって維持されているのか」
という実務的な課題意識に基づいた視点です。

## 2. Data Source / データソース

The analysis uses official statistics from the
Crop Statistics Survey (Fruit Trees) published by
the Ministry of Agriculture, Forestry and Fisheries (MAFF) of Japan.

The dataset provides prefecture-level aggregated information,
allowing for structural comparisons across regions
rather than individual farm-level analysis.

使用データは、農林水産省が公表している
**作物統計調査（果樹）**に基づいています。

本データは都道府県単位で集計された統計情報であり、
個別農家レベルではなく、
産地全体の生産構造を把握することを目的としています。

## 3. Key Variables / 使用変数

The following core variables are used consistently throughout the analysis:

Cultivated Area（結果樹面積）
Total area of bearing pear trees (ha)

Yield per 10a（10a当たり収量）
Production efficiency measured as kilograms per 10a

Production Volume（収穫量）
Total harvested volume (tons)

(Interpretation)
Production volume can be conceptually decomposed into
cultivated area and yield efficiency.
This decomposition forms the analytical foundation of the project.

（解釈）
収穫量は概念的に
「結果樹面積 × 単位収量」
として捉えることができます。

本プロジェクトでは、この分解を分析の出発点とし、
どの要素が生産を主に支えているのかを検討します。

## 4. Time Scope Selection / 分析期間の考え方

EN
Two reference years are selected for analysis: 2003 and 2024.

Rather than performing a full year-by-year time series analysis,
this study focuses on long-term structural comparison.
This approach reduces the influence of short-term weather variability
and highlights persistent changes in production structures.

JP
分析期間として 2003年と2024年 の2時点を採用しています。

年次ごとの詳細な時系列分析ではなく、
長期的な構造変化に焦点を当てることで、
天候などによる短期的な変動の影響を抑え、
生産構造そのものの変化を明確に捉えることを目的としています。

## 5. Analytical Positioning / 本分析の位置づけ

This project is positioned as a structural and exploratory analysis.
Rather than aiming to predict future production levels,
the objective is to examine how production structures,
defined by the relationship between cultivated area and yield per 10a,
have evolved across regions over time.

By organizing and visualizing these structural relationships,
the analysis provides a foundation for understanding regional differences
and long-term trends in Japanese pear production,
which may inform future policy discussions or technological considerations.

本分析は、将来の生産量を予測することを目的とするものではなく、
結果樹面積と10a当たり収量の関係性に着目した
生産構造の整理および可視化を行う探索的分析として位置づけている。

地域ごとの構造的な違いや長期的な変化を明らかにすることで、
今後の施策検討や技術導入を考える際の
基礎的な視点を提供することを意図している。
